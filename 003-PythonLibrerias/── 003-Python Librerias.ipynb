```
#Ejercicio 2 – Dataset simple con Pandas

import pandas as pd

datos = {
    "estudiante": ["Ana", "Juan", "Carla", "Pedro", "Lucia"],
    "nota": [8, 6, 9, 5, 7]
}

df = pd.DataFrame(datos)

# Mostrar
print(df)

# Calcular el promedio

promedio = df["nota"].mean()
print("Promedio de las notas:", promedio)

# Mostrar la nota máxima
maxima = df["nota"].max()
print("Nota máxima:", maxima)

# Mostrar la nota mínima
minima = df["nota"].min()
print("Nota mínima:", minima)
```
